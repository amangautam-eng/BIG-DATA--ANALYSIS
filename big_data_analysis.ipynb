```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, count, desc
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Create Spark session
spark = SparkSession.builder.appName("NYC Taxi Analysis").getOrCreate()
```

```python
# Load sample dataset (replace path with actual large CSV if available)
df = spark.read.csv("https://raw.githubusercontent.com/datablist/sample-csv-files/main/files/people/people-100.csv", header=True, inferSchema=True)
df.printSchema()
df.show(5)
```

```python
# Filter valid entries
df_clean = df.filter(col("Age").isNotNull() & (col("Age") > 20))
```

```python
# Group by age and count
age_dist = df_clean.groupBy("Age").agg(count("Age").alias("Count")).orderBy(desc("Count"))
age_dist.show(10)
```

```python
# Convert to pandas for plotting
pdf = age_dist.toPandas()

plt.figure(figsize=(10,6))
plt.bar(pdf['Age'], pdf['Count'], color='skyblue')
plt.xlabel('Age')
plt.ylabel('Number of People')
plt.title('Age Distribution')
plt.grid(True)
plt.show()
```
